## Actividad 3 — Alta cardinalidad (25 pts)

```python
# 3.1 — Calcula .nunique() para TODAS las columnas del dataset (no solo las categóricas)
# y la razón (valores únicos / total de filas) para cada una.

for columna in df.columns:
  unicos = df[columna].nunique()
  razon = unicos/len(df[columna])
  print(f'{columna:19s} ->     {unicos:5d}  valores únicos ({razon:.3%} del total de filas), ')
```

Como podemos observar, las columnas con mayor cardinalidad son las "customerID", "TotalCharges" y "MonthlyCharges". Son éstas las que no usaría como variables predictoras porque no simplifican nada la información ni se puede agrupar por ellas obteniendo un significado agregado. **3.3 — Agrupación "Top 10 + Otros"**

En clase agrupaste `country` de Netflix Titles en sus 10 categorías más frecuentes + `'Otros'`, reduciendo su cardinalidad. Aplica la misma técnica aquí sobre la columna de mayor cardinalidad que identificaste en 3.1 (pista: `.value_counts().head(10)`, luego `.where()` + `.isin()`, igual que en el notebook de clase).

```python
# Aplica el agrupamiento Top 10 + Otros sobre la columna de mayor cardinalidad
top_10 = df["customerID"].value_counts().head(10)

#Guardamos la lista de las categorías que sí se conservan
categorias_top_10 = top_10.index.tolist()              #.index lo que hace es obtener las categorías, y .tolist() lo hace una lista
print(categorias_top_10)

df["customerID_agrupado"] = df["customerID"].where(df["customerID"].isin(categorias_top_10), "Otros") #cualquier valor que no esté en categorias_top_10 se reemplaza por "otros"
df["customerID_agrupado"].value_counts()
```

## Actividad 4 — Tipos de dato (30 pts)

```python
df['TotalCharges'].dtype
```

**4.1 — Investiga (responde en Markdown):**

`TotalCharges` contiene valores numéricos (montos en dólares), pero pandas la cargó como `object`, no como `float`. Investiga por qué — revisa si hay algún valor que no se vea como un número normal.

_Tu respuesta:_
Si corremos el código **df[df["TotalCharges"]==" "]** podemos observar que hay valores vacíos, puede ser esa la razón por la que se toma como string por lo que lo toma como "object".

```python
# 4.2 — Corrige el tipo de TotalCharges.
# Pista: pd.to_numeric() con el parámetro errors= te puede ayudar a identificar
# o manejar los valores problemáticos que encontraste en 4.1.
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors = "coerce")
print("La columna ahora es formato:", df["TotalCharges"].dtype)      #Nos cersioramos de que haya cambiado
```

```python
# 4.3 — Convierte a category las columnas categóricas que, según lo que calculaste
# en la Actividad 3, tengan cardinalidad baja y valores fijos.
# Verifica con .dtypes que el cambio se aplicó correctamente.
columnas_convertir = []
for columna in df.columns:                      #Iteramos sobre cada una de las columnas
  if df[columna].dtype == "object":             #Si el tipo de la columna es object (así maneja pandas los datos categóricos)
    unicos = df[columna].nunique()              #Le contamos el número de valores únicos que tiene
    razon = unicos/len(df[columna])             #Dividimos el número de valores únicos entre el número de filas de la columna
    if razon < .1:                              #La regla para decidir que una columna tiene cardinalidad baja es que su cardinalidad sea menor a 10, o sea menor a .1%
      columnas_convertir.append(columna)

df[columnas_convertir] = df[columnas_convertir].astype("category")
df.dtypes                                        #Verificamos que se hayan convertido en "category"
```
